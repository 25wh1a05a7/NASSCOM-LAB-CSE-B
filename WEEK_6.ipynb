**OWN PROBLEMS**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.DataFrame({
    "Department": ["HR", "IT", "IT", "Finance", "Finance", "HR", "IT"],
    "Salary": [30000, 50000, 60000, 45000, 55000, 35000, 65000],
    "Age": [25, 28, 30, 27, 32, 26, 31],
    "Experience": [1, 3, 5, 2, 6, 2, 7]
})

print(data)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
departments = ["HR", "IT", "Finance"]
employees = [20, 50, 30]

plt.bar(departments, employees)
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

```python
departments = ["HR", "IT", "Finance"]
employees = [20, 50, 30]

plt.barh(departments, employees)
plt.title("Employees by Department")
plt.xlabel("Number of Employees")
plt.ylabel("Department")
plt.show()
```

```python
departments = ["HR", "IT", "Finance"]
employees = [20, 50, 30]

plt.pie(employees, labels=departments, autopct="%1.1f%%")
plt.title("Employee Distribution")
plt.show()
```

```python
plt.scatter(data["Age"], data["Salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
plt.hist(data["Salary"], bins=5)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(data["Salary"])

plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

```python
sns.countplot(x="Department", data=data)

plt.title("Number of Employees by Department")
plt.show()
```

```python
sns.violinplot(x="Department", y="Salary", data=data)

plt.title("Salary Distribution by Department")
plt.show()
```

```python
sns.swarmplot(x="Department", y="Salary", data=data)

plt.title("Individual Salary Values")
plt.show()
```

```python
sns.stripplot(x="Department", y="Salary", data=data)

plt.title("Salary Values by Department")
plt.show()
```

```python
correlation = data[["Salary", "Age", "Experience"]].corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```
