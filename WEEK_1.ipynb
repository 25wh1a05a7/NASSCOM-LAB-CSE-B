FUNCTIONS

```python
def greet():
    print("Hello! Welcome to Python Functions.")
greet()
```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Nainika")
greet_user("Bob")
```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
def square(n):
    return n*n

print("Square of 4:", square(4))
```

```python
from functools import reduce
numbers = list(map(int, input("Enter the list of elements: ").split()))
squared = list(map(lambda x: x ** 2, numbers))
even_numbers = list(filter(lambda x: x % 2 == 0, squared))
sum_even = reduce(lambda x, y: x + y, even_numbers, 0)

print(f"Original List : {numbers}")
print(f"Squared List  : {squared}")
print(f"Even Squares  : {even_numbers}")
print(f"Sum of Even Squares : {sum_even}")
```

LISTS, TUPLES, SETS AND DICTIONARIES

```python
fruits = ['apple', 'banana', 'cherry', 'apple']
print("Original list:", fruits)
print("First fruit:", fruits[0])
```

```python
fruits.append('orange')
print("After appending:", fruits)
```

```python
fruits.remove('banana')
print("After removing banana:", fruits)
print("Number of fruits:", len(fruits))
```

```python
dimensions = (10, 20, 30)
print("Tuple:", dimensions)
```

```python
print("Second element:", dimensions[1])
```

```python
unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)
```

```python
unique_numbers.add(4)
print("After adding 4:", unique_numbers)
```

```python
unique_numbers.remove(2)
print("After removing 2:", unique_numbers)
```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```
