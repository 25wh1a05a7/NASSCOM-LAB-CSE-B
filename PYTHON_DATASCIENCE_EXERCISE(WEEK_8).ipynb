**1. Setup & Dataset Initialization**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                    111, 112, 113, 114, 115],
    "Department": ["IT", "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "IT", "HR", "Finance",
                   "IT", "HR", "Finance", "IT", "HR"],
    "Age": [22, 25, 28, 30, 24, 35, 40, 27, 29, 45,
            32, 26, 38, 31, 50],
    "Experience": [1, 2, 4, 6, 2, 10, 15, 3, 5, 18,
                   8, 3, 12, 7, 20],
    "Salary": [25000, 28000, 32000, 40000, 27000,
               55000, 80000, 35000, 38000, 95000,
               60000, 30000, 75000, 50000, 150000],
    "Performance_Score": [65, 70, 75, 80, 68,
                          85, 90, 78, 72, 95,
                          88, 74, 92, 82, 96],
    "Training_Completed": ["Yes", "No", "Yes", "Yes", "No",
                           "Yes", "Yes", "No", "Yes", "Yes",
                           "Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)
print("--- Full Dataset ---")
print(df)
```

**2. Part A – Understanding Data Types**

```python
# First 5 records
print("--- First 5 Records ---")
print(df.head())

# Last 5 records
print("\n--- Last 5 Records ---")
print(df.tail())

# Number of rows and columns
print("\nShape (Rows, Columns):", df.shape)

# Column names
print("\nColumn Names:", df.columns.tolist())

# Data types
print("\n--- Data Types ---")
print(df.dtypes)
```

**3. Part B – Basic Data Exploration**

```python
# Descriptive statistics
print("--- Descriptive Statistics ---")
print(df.describe())

# Department distribution
print("\n--- Department Count ---")
print(df["Department"].value_counts())

# Training completed distribution
print("\n--- Training Completed Count ---")
print(df["Training_Completed"].value_counts())

# Check missing values
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

**4. Part C – Measures of Central Tendency**

```python
num_cols = ["Age", "Experience", "Salary", "Performance_Score"]

# Calculate Mean, Median, and Mode
mean_vals = df[num_cols].mean()
median_vals = df[num_cols].median()

print("--- Mean ---")
print(mean_vals)

print("\n--- Median ---")
print(median_vals)

print("\n--- Mode ---")
for col in num_cols:
    print(f"{col} Mode:", df[col].mode().tolist())

central_tendency_df = pd.DataFrame({
    "Mean": mean_vals,
    "Median": median_vals,
    "Mode": [df[col].mode().tolist() for col in num_cols]
}, index=num_cols)

print("\n--- Part C Table Summary ---")
print(central_tendency_df)
```

**5. Part D – Measures of Spread / Dispersion**

```python
# Calculate Range.....
spread_summary = []

for col in num_cols:
    minimum = df[col].min()
    maximum = df[col].max()
    rng = maximum - minimum
    variance = df[col].var()
    std_dev = df[col].std()
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    spread_summary.append({
        "Variable": col,
        "Min": minimum,
        "Max": maximum,
        "Range": rng,
        "Variance": round(variance, 2),
        "Std. Dev.": round(std_dev, 2),
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr
    })

spread_df = pd.DataFrame(spread_summary).set_index("Variable")
print("--- Part D Table Summary ---")
print(spread_df)
```

**6. Part E – Skewness**

```python
skewness_vals = df[num_cols].skew()

interpretations = []
for val in skewness_vals:
    if val > 0.5:
        interpretations.append("Positively Skewed (Right-Skewed)")
    elif val < -0.5:
        interpretations.append("Negatively Skewed (Left-Skewed)")
    else:
        interpretations.append("Approximately Symmetric")

skew_df = pd.DataFrame({
    "Skewness": round(skewness_vals, 2),
    "Interpretation": interpretations
}, index=num_cols)

print("--- Part E Skewness Summary ---")
print(skew_df)
```

**7. Part F – Visualization**

```python
# A. Histogram
plt.figure(figsize=(8, 4))
plt.hist(df["Salary"], bins=6, color="skyblue", edgecolor="black")
plt.xlabel("Salary (₹)")
plt.ylabel("Frequency")
plt.title("Salary Distribution Histogram")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# B. Box Plot
plt.figure(figsize=(6, 4))
plt.boxplot(df["Salary"], patch_artist=True, boxprops=dict(facecolor="lightgreen"))
plt.ylabel("Salary (₹)")
plt.title("Salary Box Plot (Outlier Detection)")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**8. Part G – Combined Statistical Summary**

```python
summary = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Std_Deviation": df[num_cols].std(),
    "Skewness": df[num_cols].skew()
}, index=num_cols)

print("--- Combined Statistical Summary ---")
print(summary.round(2))
```

**9. Part H – Outlier Analysis & Impact**

```python
df_no_outlier = df[df["Salary"] != 150000]

print("--- Before Removing Outlier (₹150,000) ---")
print(f"Mean Salary: ₹{df['Salary'].mean():.2f}")
print(f"Median Salary: ₹{df['Salary'].median():.2f}")
print(f"Std Deviation: ₹{df['Salary'].std():.2f}")
print(f"Skewness: {df['Salary'].skew():.2f}")

print("\n--- After Removing Outlier (₹150,000) ---")
print(f"Mean Salary: ₹{df_no_outlier['Salary'].mean():.2f}")
print(f"Median Salary: ₹{df_no_outlier['Salary'].median():.2f}")
print(f"Std Deviation: ₹{df_no_outlier['Salary'].std():.2f}")
print(f"Skewness: {df_no_outlier['Salary'].skew():.2f}")
```

**10. Part I – Group-Based Analysis**

```python
dept_mean = df.groupby("Department")["Salary"].mean()
dept_median = df.groupby("Department")["Salary"].median()
dept_std = df.groupby("Department")["Salary"].std()

dept_summary = pd.DataFrame({
    "Mean Salary": dept_mean,
    "Median Salary": dept_median,
    "Std Dev (Variation)": dept_std
})

print("--- Department-wise Salary Analysis ---")
print(dept_summary.round(2))
```

**11. Part J – Automated Statistical Pipeline (Final Challenge)**

```python
numeric_columns = df.select_dtypes(include=np.number).columns
print("   AUTOMATED NUMERICAL STATISTICAL ANALYSIS       ")

for column in numeric_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    print(f"\nCOLUMN: {column}")
    print("-" * 30)
    print(f"Mean               : {df[column].mean():.2f}")
    print(f"Median             : {df[column].median():.2f}")
    print(f"Mode               : {df[column].mode().tolist()}")
    print(f"Minimum            : {df[column].min()}")
    print(f"Maximum            : {df[column].max()}")
    print(f"Range              : {df[column].max() - df[column].min()}")
    print(f"Variance           : {df[column].var():.2f}")
    print(f"Standard Deviation : {df[column].std():.2f}")
    print(f"Q1 (25th percentile): {q1}")
    print(f"Q3 (75th percentile): {q3}")
    print(f"IQR                : {iqr}")
    print(f"Skewness           : {df[column].skew():.2f}")
```
